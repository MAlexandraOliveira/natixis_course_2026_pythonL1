## 4. Basic syntax: `print()`, comments, and program flow

### `print()` — show a result

`print()` displays whatever you give it: text, a number, or the value stored in a variable.
It is how you see what your program is doing.

### Comments — `#`

Anything after a `#` is a **comment**. Python ignores it completely — it is a note for
humans. Good comments explain *why*, not *what*.

### Program flow — one line at a time

Python runs your code **top to bottom**, like a recipe. Order matters:

```
   total = 100
       │
       ▼
   print(total)  ──────▶  100
       │
       ▼
   total = total + 50
       │
       ▼
   print(total)  ──────▶  150
```

The second `print()` shows a different number because the line above it changed `total`.

````python
# A comment: this line is ignored by Python
print("My first Python program")
print("The answer is:", 42)

total = 100
print(total)          # shows 100

total = total + 50    # this line changes total
print(total)          # shows 150 - the order of the lines matters
````

## 5. Variables and data types

### Variables

A **variable** is a named box that holds a value. You create one by writing `name = value`.

| ✅ Valid | ❌ Invalid | Why |
|----------|-----------|-----|
| `total` | `2nd_total` | cannot start with a number |
| `total_paid` | `total paid` | no spaces |
| `clientName` | `client-name` | no hyphens |
| `TOTAL` | `if` | reserved Python word |

The convention in Python is `lower_case_with_underscores`.

### Data types

Every value has a **type**, which decides what you can do with it. Today we use five:

| Type | Keyword | Example | Used for |
|------|---------|---------|----------|
| Text | `str` | `"Ana Silva"` | names, labels, codes |
| Whole number | `int` | `34` | counts, quantities |
| Decimal | `float` | `149.90` | prices, rates, averages |
| True / False | `bool` | `True` | flags, checks |
| A sequence | `list` | `["Porto", "Lisbon"]` | several values at once |

`type()` tells you the type of any value, and `len()` how many items a list holds.

> 💡 There is a sixth type you will use constantly — the **dictionary** (`dict`), which
> stores values under names. That is the first topic of Class 2.

````python
# One customer, stored across several variables
name       = "Ana Silva"                    # str   - text
age        = 34                             # int   - a whole number
total_paid = 149.90                         # float - a decimal number
is_active  = True                           # bool  - True or False
cities     = ["Porto", "Lisbon", "Faro"]    # list  - several values

print(name, age, total_paid)
print(type(name))         # <class 'str'>
print(type(age))          # <class 'int'>
print(type(total_paid))   # <class 'float'>
print(type(is_active))    # <class 'bool'>
print(type(cities))       # <class 'list'>
print(len(cities))        # 3 - how many items in the list
````

## 6. Casting and `input()`

### Casting — changing a value's type

Sometimes a number arrives as text — for example `"34"` read from a file. You cannot do
arithmetic with text, so you **cast** it: `int()`, `float()` or `str()`.

### `input()` — asking the user

`input()` pauses the program and waits for someone to type. Whatever they type comes back
as **text**, always — so if you need a number, you must cast it.

> ⚠️ Forgetting to cast after `input()` is one of the most common beginner bugs. `"10" + "5"`
> gives `"105"`, not `15`.

````python
age_text = "34"            # this is TEXT, not a number
print(type(age_text))      # <class 'str'>

age = int(age_text)        # cast it to a whole number
print(type(age))           # <class 'int'>
print(age + 1)             # now arithmetic works: 35

# input() always returns text - cast it before calculating
typed = input("How many items did you buy? ")
items = int(typed)
print("Total price:", items * 2.50)
````

## 7. Operators

### Arithmetic

| Operator | Name | Example | Result |
|----------|------|---------|--------|
| `+` | Add | `3 + 2` | `5` |
| `-` | Subtract | `3 - 2` | `1` |
| `*` | Multiply | `3 * 2` | `6` |
| `/` | Divide | `7 / 2` | `3.5` |
| `//` | Whole division | `7 // 2` | `3` |
| `%` | Remainder | `7 % 2` | `1` |
| `**` | Power | `2 ** 3` | `8` |

`round(value, 2)` rounds a number to 2 decimal places.

### Assignment shortcuts

`+=` and `-=` update a variable using its own value. You will use `+=` constantly to count
things inside a loop.

| Shortcut | Same as |
|----------|---------|
| `x += 3` | `x = x + 3` |
| `x -= 3` | `x = x - 3` |

### Comparison operators — always return `True` or `False`

| Operator | Meaning | Example |
|----------|---------|--------|
| `==` | Equal | `x == 5` |
| `!=` | Not equal | `x != 5` |
| `>` | Greater than | `x > 5` |
| `<` | Less than | `x < 5` |
| `>=` | Greater or equal | `x >= 5` |
| `<=` | Less or equal | `x <= 5` |

> ⚠️ `=` **assigns** a value. `==` **compares** two values. Mixing them up is a classic bug.

### Logical — combining conditions

| Operator | Meaning | Example |
|----------|---------|--------|
| `and` | Both must be True | `x > 0 and x < 10` |
| `or` | At least one True | `x < 0 or x > 10` |
| `not` | Reverses the result | `not (x == 5)` |

### Membership — `in` and `not in`

Always return `True` or `False`.

| Operator | Meaning | Example |
|----------|---------|--------|
| `in` | is this value somewhere in that list? | `x in list_values` |
| `not in` | is this value not in that list? | `x not in list_values` |


````python
price = 2.50
quantity = 4

total = price * quantity
print("Total:", total)              # 10.0
print(round(10.567, 2))             # 10.57

# // and % answer "how many fit?" and "what is left over?"
print(7 // 2, 7 % 2)                # 3 1

count = 0
count += 1                          # same as: count = count + 1
count += 1
print("Count:", count)              # 2

# Comparison gives True or False
print(total > 5)                    # True
print(total == 0)                   # False

# Logical operators combine conditions
print(total > 5 and quantity > 2)   # True
print(not total > 5)                # False

# Membership - is a value in this list?
cities = ["Porto", "Lisbon", "Faro"]
print("Madrid" in cities)           # False
print("Madrid" not in cities)       # True
````

## 8. f-strings — putting values inside text

Gluing text and numbers together with `+` is painful: every number has to be converted with
`str()` first, and the result is hard to read.

An **f-string** is an ordinary piece of text with an `f` before the opening quote. Anything
you write inside `{}` is replaced by its value:

`print(f"{name} paid {total} EUR")`

### Format specifiers — controlling how a value is shown

A **format specifier** goes after a colon inside the `{}`, and controls how the value is
displayed. This is how you make numbers look presentable:

| Specifier | Meaning | Example | Result |
|-----------|---------|---------|--------|
| `:.2f` | 2 decimal places | `f"{3.14159:.2f}"` | `3.14` |
| `:.1f` | 1 decimal place | `f"{7.8341:.1f}"` | `7.8` |
| `:,` | thousands separator | `f"{1234567:,}"` | `1,234,567` |
| `:,.2f` | both together | `f"{1234.5:,.2f}"` | `1,234.50` |
| `:>8` | right-align within 8 characters | `f"{42:>8}"` | six spaces, then `42` |
| `:<8` | left-align within 8 characters | `f"{'Tea':<8}"` | `Tea`, then five spaces |

The two alignment specifiers are what make columns line up in a printed list — you will use
them in section 9.

> 💡 Use f-strings everywhere. They are the standard way to build output in modern Python, and
> you will see them in every remaining class.

````python
name = "Ana Silva"
total = 1234.5

# The old, painful way - every number needs str()
print("Customer " + name + " paid " + str(total) + " EUR")

# The same thing as an f-string: shorter, readable, no str() needed
print(f"Customer {name} paid {total} EUR")

# Format specifiers make numbers presentable
print(f"Total: {total:.2f} EUR")        # Total: 1234.50 EUR
print(f"Total: {total:,.2f} EUR")       # Total: 1,234.50 EUR

# Alignment lines the columns up
print(f"{'Coffee':<10}{8.50:>8.2f}")
print(f"{'Tea':<10}{4.20:>8.2f}")
````

## 9. Making decisions — `if` / `elif` / `else`

`if` lets your program choose what to do. `elif` ("else if") tries another condition, and
`else` catches everything left over. Python checks the conditions **in order** and stops at
the first one that is `True`.

For example, deciding what to do about an order, based on its status:

```
       ╱──────────────╲
      ╱   status ==     ╲── Yes ──▶ "nothing to do"
      ╲    "paid" ?     ╱
       ╲──────────────╱
             │ No
             ▼
       ╱──────────────╲
      ╱   status ==     ╲── Yes ──▶ "refund the customer"
      ╲  "cancelled" ?  ╱
       ╲──────────────╱
             │ No
             ▼
      "send a reminder"
```

> ⚠️ **Indentation is part of the language.** The lines belonging to an `if` must be indented
> (4 spaces, or one Tab). Python uses that indentation to know which lines belong to the
> block — there are no brackets. Get it wrong and the program does the wrong thing.

````python
status = "cancelled"

if status == "paid":
    print("Nothing to do")
elif status == "cancelled":
    print("Refund the customer")
else:
    print("Send a payment reminder")

# Conditions can be combined
total = 1250.00
if total > 1000 and status != "paid":
    print("Large unpaid order - call the customer")
````

## 10. Repeating work — `while`, `for`, `break`, `continue`

Doing the same thing to 400 rows is what computers are for. A **loop** repeats work. Python
has two, and they answer different questions.

### `while` — repeat as long as something is true

Use `while` when you don't know in advance how many times you will go round. The arrow back
to the test is the important part:

```
        items_left = 3
              │
              ▼
       ╱────────────────╲
      ╱  items_left > 0 ? ╲── No ──▶ "All checked"
      ╲                  ╱             (loop ends)
       ╲────────────────╱
              │ Yes
              ▼
       print(items_left)
              │
              ▼
       items_left = items_left - 1
              │
              └────────▶ back to the test
```

> ⚠️ Something inside the loop must eventually make the condition `False`, or it runs
> forever. In Colab, press the ■ stop button if that happens.

### `for` — do something once per item

Use `for` when you know what you are working through: every item in a list, or a fixed number
of repetitions. You don't manage a counter — Python does it for you. `range()` generates
numbers to loop over:

- `range(4)` → `0, 1, 2, 3`
- `range(1, 4)` → `1, 2, 3`
- `range(0, 10, 2)` → `0, 2, 4, 6, 8`

### `break` and `continue` — changing the loop's course

| Keyword | Effect | Use it when |
|---------|--------|-------------|
| `break` | leave the loop **completely**, immediately | you have found what you were looking for |
| `continue` | skip the rest of **this one turn**, go to the next item | this item is not relevant |

The difference is the whole point: `break` abandons the entire loop, `continue` abandons only
the current turn and carries on with the next item.

````python
# while - repeat as long as the condition is True
items_left = 3
while items_left > 0:
    print(f"{items_left} item(s) left to check...")
    items_left -= 1           # without this line, the loop would never end
print("All checked")

# for - once per item in the list
cities = ["Porto", "Lisbon", "Faro"]
for city in cities:
    print(f"Visiting {city}")

# range() - a fixed number of times, and break to stop early
for attempt in range(1, 11):
    print(f"Attempt {attempt}")
    if attempt == 3:
        print("Found it - no need to try the rest")
        break                 # leaves the loop completely

# continue - skip this one item, carry on with the rest
prices = [8.50, 0.00, 4.20]
for price in prices:
    if price == 0:
        print("Free item - skipping")
        continue              # jump straight to the next price
    print(f"Charging {price:.2f} EUR")
````

## 11. 🔗 Putting it all together — invoice review

Everything from today, in one program.

### The situation

You have a list of **invoice amounts** to review before they get paid. Small ones can be paid
straight away, big ones need a manager's approval, and a cancelled invoice has an amount of
`0` and should be ignored. Doing this by hand is slow and easy to get wrong.

### What the program must do

1. Work through the list of amounts, one at a time
2. Give each invoice a reference number, counting up from `INV-417`
3. **Skip** any cancelled invoice — one with an amount of `0` — and move on to the next
4. Decide what each remaining invoice needs:
   - over 1000 EUR → `needs approval`
   - under 100 EUR → `pay automatically`
   - anything in between → `normal check`
5. Print one tidy line per invoice, with the columns lined up
6. At the end, print how many were **reviewed**, how many **need approval**, and the **total
   value** — then say whether there is an approval list to send

### Where each concept appears

| Concept from today | Where it is used |
|--------------------|------------------|
| Variables, `float`, `str` | the amounts, the two limits, the reference |
| `list` | the list of amounts |
| `for` loop | one turn per invoice |
| `continue` | skipping the cancelled invoice |
| `if` / `elif` / `else` | choosing what each invoice needs |
| Comparison operators | `> approval_limit`, `< small_limit`, `== 0` |
| `+=` | the three running totals and the reference counter |
| f-strings with specifiers | the lined-up columns |
| Comments | explaining each block |

````python
# ── The invoices to review, in EUR ─────────────────────────────────────────
amounts = [1250.00, 87.50, 0.00, 640.00, 2400.00]

# ── The rules ──────────────────────────────────────────────────────────────
approval_limit = 1000.00   # above this, a manager must approve
small_limit = 100.00       # below this, it is paid automatically

next_number = 417          # ── running totals, all starting from zero ──
reviewed = 0
needs_approval = 0
total_value = 0.00

print("INVOICE REVIEW")
print("-" * 46)

# ── One turn per invoice ───────────────────────────────────────────────────
for amount in amounts:
    reference = f"INV-{next_number}"
    next_number += 1

    # A cancelled invoice has no amount - skip it and take the next one
    if amount == 0:
        print(f"{reference}   cancelled - skipped")
        continue

    reviewed += 1
    total_value += amount

    # Decide what this invoice needs
    if amount > approval_limit:
        action = "needs approval"
        needs_approval += 1
    elif amount < small_limit:
        action = "pay automatically"
    else:
        action = "normal check"

    print(f"{reference}   {amount:>9,.2f} EUR   {action}")

print("-" * 46)             # ── the summary ──
print(f"Invoices reviewed : {reviewed}")
print(f"Needing approval  : {needs_approval}")
print(f"Total value       : {total_value:,.2f} EUR")

if needs_approval > 0:
    print("\nSend the approval list to the manager.")
else:
    print("\nNothing needs approval today.")
````
